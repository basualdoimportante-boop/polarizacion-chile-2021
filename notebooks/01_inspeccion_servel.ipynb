```python
# Copia de trabajo
df_piloto = df.copy()

# 1. Eliminar filas basura (sin comuna, sin distrito, sin mesa)
print("Filas iniciales:", len(df_piloto))
df_piloto = df_piloto.dropna(subset=['Comuna', 'Distrito', 'Nro. Mesa'])
print("Filas tras eliminar basura:", len(df_piloto))

# 2. Eliminar filas sin votos
df_piloto = df_piloto.dropna(subset=['Votos TRICEL'])
print("Filas tras eliminar votos nulos:", len(df_piloto))

# 3. Separar votos nulos/blancos
mask_especiales = df_piloto['Candidato'].isin(['VOTOS NULOS', 'VOTOS EN BLANCO'])
df_especiales = df_piloto[mask_especiales].copy()
df_validos = df_piloto[~mask_especiales].copy()
print("Filas votos nulos/blancos (separadas):", len(df_especiales))
print("Filas votos válidos:", len(df_validos))

# 4. Crear columna 'Bloque'
df_validos['Bloque'] = df_validos['Pacto'].fillna(df_validos['Partido'])
print("\nBloques únicos (conteo de filas):")
print(df_validos['Bloque'].value_counts())
```

```python
# 1. Agregar votos por Comuna y Bloque
df_agregado = df_validos.groupby(['Comuna', 'Bloque'])['Votos TRICEL'].sum().reset_index()

# 2. Ver la tabla agregada
print("=== VOTOS POR COMUNA Y BLOQUE (primeras filas) ===")
print(df_agregado.head(20))

# 3. Verificar que la suma de votos válidos coincide
print("\n=== VERIFICACIÓN DE TOTALES ===")
print("Total de votos en dataset original:", df_validos['Votos TRICEL'].sum())
print("Total de votos en dataset agregado:", df_agregado['Votos TRICEL'].sum())

# 4. Calcular la cuota de cada bloque por comuna (proporción)
df_agregado['Cuota'] = df_agregado.groupby('Comuna')['Votos TRICEL'].transform(lambda x: x / x.sum())

print("\n=== CUOTAS POR BLOQUE (primeras filas) ===")
print(df_agregado.head(20))

# 5. Calcular un índice de fragmentación simple por comuna (número efectivo de bloques)
# Este es un primer paso para medir polarización
def numero_efectivo_bloques(cuotas):
    # Índice de Laakso-Taagepera: 1 / sum(cuota^2)
    return 1 / (cuotas ** 2).sum()

fragmentacion = df_agregado.groupby('Comuna')['Cuota'].apply(numero_efectivo_bloques).reset_index()
fragmentacion.columns = ['Comuna', 'Numero_Efectivo_Bloques']

print("\n=== NÚMERO EFECTIVO DE BLOQUES POR COMUNA (primeras 15) ===")
print(fragmentacion.sort_values('Numero_Efectivo_Bloques', ascending=False).head(15))
```

```python
import glob
import pandas as pd

# Listar archivos
archivos = sorted(glob.glob("../datos/Diputados/*.xlsx"))
print(f"Archivos a procesar: {len(archivos)}")

# Lista para acumular DataFrames
dataframes = []
resumen = []

for archivo in archivos:
    try:
        # Leer todas las hojas del archivo
        xls = pd.ExcelFile(archivo)
        hojas = xls.sheet_names
        
        for hoja in hojas:
            df_temp = pd.read_excel(archivo, sheet_name=hoja)
            # Añadir columnas de trazabilidad
            df_temp['archivo_origen'] = archivo.split('\\')[-1]
            df_temp['hoja_origen'] = hoja
            dataframes.append(df_temp)
            resumen.append({
                'archivo': archivo.split('\\')[-1],
                'hoja': hoja,
                'filas': len(df_temp)
            })
        
        print(f"OK: {archivo.split(chr(92))[-1]} ({len(hojas)} hojas)")
    except Exception as e:
        print(f"ERROR en {archivo}: {e}")

# Concatenar todo
df_nacional = pd.concat(dataframes, ignore_index=True)

print("\n=== RESUMEN DE CONCATENACIÓN ===")
print(f"Total de archivos procesados: {len(archivos)}")
print(f"Total de hojas leídas: {len(dataframes)}")
print(f"Filas totales: {len(df_nacional)}")
print(f"Columnas: {len(df_nacional.columns)}")

# Verificación de columnas
print("\nColumnas del dataset nacional:")
print(df_nacional.columns.tolist())

# Ver distribución por región
print("\n=== FILAS POR REGIÓN ===")
print(df_nacional.groupby('Región').size().sort_values(ascending=False))
```

```python
# 1. Limpieza sobre el dataset nacional
df_nac = df_nacional.copy()
print("Filas iniciales:", len(df_nac))

# Eliminar filas basura
df_nac = df_nac.dropna(subset=['Comuna', 'Distrito', 'Nro. Mesa'])
print("Filas tras eliminar basura:", len(df_nac))

# Eliminar votos nulos
df_nac = df_nac.dropna(subset=['Votos TRICEL'])
print("Filas tras eliminar votos nulos:", len(df_nac))

# Separar votos nulos/blancos
mask_esp = df_nac['Candidato'].isin(['VOTOS NULOS', 'VOTOS EN BLANCO'])
df_esp_nac = df_nac[mask_esp].copy()
df_val_nac = df_nac[~mask_esp].copy()
print("Votos nulos/blancos (separados):", len(df_esp_nac))
print("Votos válidos:", len(df_val_nac))

# Crear columna Bloque
df_val_nac['Bloque'] = df_val_nac['Pacto'].fillna(df_val_nac['Partido'])

# 2. Agregar por Región, Comuna, Distrito y Bloque
# OJO: la columna es 'Región' con tilde
df_agr_nac = df_val_nac.groupby(['Región', 'Comuna', 'Distrito', 'Bloque'])['Votos TRICEL'].sum().reset_index()

# 3. Verificación
print("\n=== VERIFICACIÓN DE TOTALES ===")
print("Votos válidos en dataset nacional:", df_val_nac['Votos TRICEL'].sum())
print("Votos en dataset agregado:", df_agr_nac['Votos TRICEL'].sum())
print("¿Coinciden?", df_val_nac['Votos TRICEL'].sum() == df_agr_nac['Votos TRICEL'].sum())

# 4. Dimensiones del dataset agregado
print("\n=== DIMENSIONES DEL DATASET AGREGADO ===")
print(f"Filas: {len(df_agr_nac)}")
print(f"Comunas únicas: {df_agr_nac['Comuna'].nunique()}")
print(f"Regiones únicas: {df_agr_nac['Región'].nunique()}")
print(f"Bloques únicos: {df_agr_nac['Bloque'].nunique()}")
print("\nBloques:")
print(df_agr_nac['Bloque'].unique())

# 5. Primeras filas
print("\n=== PRIMERAS FILAS ===")
print(df_agr_nac.head(20))
```

```python
# 1. Filtrar solo bloques con votos > 0 (excluir los que no compitieron)
df_agr_pos = df_agr_nac[df_agr_nac['Votos TRICEL'] > 0].copy()

print("=== FILAS CON VOTOS > 0 ===")
print(f"Filas: {len(df_agr_pos)}")
print(f"Comunas: {df_agr_pos['Comuna'].nunique()}")

# 2. Calcular total de votos válidos por comuna
total_por_comuna = df_agr_pos.groupby('Comuna')['Votos TRICEL'].sum().reset_index()
total_por_comuna.columns = ['Comuna', 'Votos_Validos_Comuna']

# 3. Merge para tener el total en cada fila
df_agr_pos = df_agr_pos.merge(total_por_comuna, on='Comuna', how='left')

# 4. Calcular cuota (proporción) de cada bloque dentro de la comuna
df_agr_pos['Cuota'] = df_agr_pos['Votos TRICEL'] / df_agr_pos['Votos_Validos_Comuna']

# 5. Verificación: las cuotas deben sumar 1 por comuna
verif = df_agr_pos.groupby('Comuna')['Cuota'].sum()
print("\n=== VERIFICACIÓN: SUMA DE CUOTAS POR COMUNA ===")
print(f"Mínimo: {verif.min()}")
print(f"Máximo: {verif.max()}")
print(f"¿Todas suman 1? {(verif.round(6) == 1).all()}")

# 6. Primeras filas
print("\n=== PRIMERAS FILAS ===")
print(df_agr_pos[['Región', 'Comuna', 'Bloque', 'Votos TRICEL', 'Cuota']].head(20))
```

```python
import numpy as np

# Función que calcula las tres métricas a partir de las cuotas de cada comuna
def metricas_fragmentacion(cuotas):
    cuotas = np.array(cuotas)
    cuotas = cuotas[cuotas > 0]  # por seguridad, solo cuotas positivas
    
    # 1. Número Efectivo de Bloques (Laakso-Taagepera)
    # 1 / sum(cuota^2). Valores: 1 = un solo bloque, N = N bloques iguales
    laakso = 1 / np.sum(cuotas ** 2)
    
    # 2. Índice de Herfindahl-Hirschman (HHI)
    # sum(cuota^2). Valores: 1 = un solo bloque, 0 = muchos bloques pequeños
    hhi = np.sum(cuotas ** 2)
    
    # 3. Entropía de Shannon (normalizada por ln(N) para comparabilidad)
    # Valores: 0 = un solo bloque, 1 = máxima diversidad
    n_bloques = len(cuotas)
    if n_bloques > 1:
        entropia = -np.sum(cuotas * np.log(cuotas)) / np.log(n_bloques)
    else:
        entropia = 0
    
    return pd.Series({
        'Laakso_Taagepera': laakso,
        'HHI': hhi,
        'Entropia_Shannon': entropia,
        'N_Bloques': n_bloques
    })

# Aplicar a cada comuna
metricas = df_agr_pos.groupby(['Región', 'Comuna'])['Cuota'].apply(metricas_fragmentacion).unstack()

print("=== MÉTRICAS DE FRAGMENTACIÓN POR COMUNA ===")
print(f"Total de comunas: {len(metricas)}")
print("\nPrimeras 15 filas ordenadas por Laakso-Taagepera (más fragmentadas primero):")
print(metricas.sort_values('Laakso_Taagepera', ascending=False).head(15))

print("\nPrimeras 15 filas ordenadas por Laakso-Taagepera (menos fragmentadas primero):")
print(metricas.sort_values('Laakso_Taagepera', ascending=True).head(15))

print("\n=== ESTADÍSTICAS DESCRIPTIVAS ===")
print(metricas.describe())
```
