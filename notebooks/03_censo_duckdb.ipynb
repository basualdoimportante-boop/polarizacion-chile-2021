```python
import pandas as pd
import unicodedata
import re

def normalizar_nombre(nombre):
    if pd.isna(nombre):
        return None
    n = str(nombre).upper().strip()
    # Quitar contenido entre paréntesis (ej: CABO DE HORNOS(EX-NAVARINO))
    n = re.sub(r'\(.*?\)', '', n)
    # Quitar tildes y diéresis
    n = ''.join(
        c for c in unicodedata.normalize('NFD', n)
        if unicodedata.category(c) != 'Mn'
    )
    # Reemplazos específicos
    n = n.replace("O'HIGGINS", "OHIGGINS")
    n = n.replace("O HIGGINS", "OHIGGINS")
    n = n.replace("MARCHIGUE", "MARCHIHUE")
    # Limpiar espacios múltiples y extremos
    n = re.sub(r'\s+', ' ', n).strip()
    return n

# Re-normalizar Censo
df_comunas['nombre_norm'] = df_comunas['Territorio'].apply(normalizar_nombre)

# Re-normalizar electoral
df_electoral['comuna_norm'] = df_electoral['Comuna'].apply(normalizar_nombre)

# Verificar coincidencias de nuevo
comunas_censo = set(df_comunas['nombre_norm'])
comunas_electoral = set(df_electoral['comuna_norm'])

print(f"Comunas en Censo: {len(comunas_censo)}")
print(f"Comunas en electoral: {len(comunas_electoral)}")
print(f"\nComunas que coinciden: {len(comunas_censo & comunas_electoral)}")

solo_censo = comunas_censo - comunas_electoral
solo_electoral = comunas_electoral - comunas_censo

if solo_censo:
    print(f"\nComunas solo en Censo ({len(solo_censo)}):")
    for c in sorted(solo_censo):
        print(f"  {c}")
else:
    print("\n✅ No hay comunas solo en Censo")

if solo_electoral:
    print(f"\nComunas solo en electoral ({len(solo_electoral)}):")
    for c in sorted(solo_electoral):
        print(f"  {c}")
else:
    print("\n✅ No hay comunas solo en electoral")
```

```python
import duckdb
import time

ruta_censo = r"C:\Users\Martín\Desktop\POLARIZACOIN\datos\censo2024\personas_censo2024.csv"

inicio = time.time()

df_censo_comunal = duckdb.sql(f"""
    WITH base AS (
        SELECT * FROM read_csv_auto('{ruta_censo}')
    ),
    religion AS (
        SELECT 
            comuna,
            SUM(CASE WHEN p31_religion = 1 THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN p31_religion IN (1,2,5,10,12,999) THEN 1 ELSE 0 END), 0) AS pct_catolicos,
            SUM(CASE WHEN p31_religion = 2 THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN p31_religion IN (1,2,5,10,12,999) THEN 1 ELSE 0 END), 0) AS pct_evangelicos,
            SUM(CASE WHEN p31_religion = 12 THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN p31_religion IN (1,2,5,10,12,999) THEN 1 ELSE 0 END), 0) AS pct_sin_religion
        FROM base
        WHERE p31_religion IN (1,2,5,10,12,999)
        GROUP BY comuna
    ),
    educacion AS (
        SELECT
            comuna,
            SUM(CASE WHEN cine11 IN (35,46,56,64) THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN cine11 NOT IN (-99, 98) THEN 1 ELSE 0 END), 0) AS pct_superior,
            SUM(CASE WHEN cine11 IN (24,25) THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN cine11 NOT IN (-99, 98) THEN 1 ELSE 0 END), 0) AS pct_secundaria,
            SUM(CASE WHEN cine11 IN (1,2,3,4,5) THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN cine11 NOT IN (-99, 98) THEN 1 ELSE 0 END), 0) AS pct_basica_o_menos,
            AVG(CASE WHEN escolaridad >= 0 THEN escolaridad END) AS promedio_escolaridad
        FROM base
        WHERE cine11 NOT IN (-99, 98)
        GROUP BY comuna
    ),
    trabajo AS (
        SELECT
            comuna,
            SUM(CASE WHEN sit_fuerza_trabajo = 2 THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN sit_fuerza_trabajo IN (1,2) THEN 1 ELSE 0 END), 0) AS tasa_desocupacion
        FROM base
        WHERE sit_fuerza_trabajo IN (1,2,3)
        GROUP BY comuna
    ),
    ruralidad AS (
        SELECT
            comuna,
            SUM(CASE WHEN area = 2 THEN 1 ELSE 0 END) * 100.0 / 
                NULLIF(SUM(CASE WHEN area IN (1,2) THEN 1 ELSE 0 END), 0) AS pct_rural
        FROM base
        WHERE area IN (1,2)
        GROUP BY comuna
    )
    SELECT 
        r.comuna,
        r.pct_catolicos, r.pct_evangelicos, r.pct_sin_religion,
        e.pct_superior, e.pct_secundaria, e.pct_basica_o_menos, e.promedio_escolaridad,
        t.tasa_desocupacion,
        ru.pct_rural
    FROM religion r
    LEFT JOIN educacion e ON r.comuna = e.comuna
    LEFT JOIN trabajo t ON r.comuna = t.comuna
    LEFT JOIN ruralidad ru ON r.comuna = ru.comuna
    ORDER BY r.comuna
""").df()

fin = time.time()
print(f"Tiempo: {fin - inicio:.2f} segundos")
print(f"\n=== RESULTADO ===")
print(f"Comunas: {len(df_censo_comunal)}")
print(f"Columnas: {df_censo_comunal.columns.tolist()}")
print("\n=== PRIMERAS 10 FILAS ===")
print(df_censo_comunal.head(10).to_string())
```

```python
import pandas as pd
import unicodedata
import re

def normalizar_nombre(nombre):
    if pd.isna(nombre):
        return None
    n = str(nombre).upper().strip()
    n = re.sub(r'\(.*?\)', '', n)
    n = ''.join(
        c for c in unicodedata.normalize('NFD', n)
        if unicodedata.category(c) != 'Mn'
    )
    n = n.replace("O'HIGGINS", "OHIGGINS")
    n = n.replace("O HIGGINS", "OHIGGINS")
    n = n.replace("MARCHIGUE", "MARCHIHUE")
    n = re.sub(r'\s+', ' ', n).strip()
    return n

# Cargar códigos territoriales para obtener el nombre de cada comuna
ruta_excel = r"C:\Users\Martín\Desktop\POLARIZACOIN\datos\censo2024\diccionario_variables_censo2024.xlsx"
df_codigos = pd.read_excel(ruta_excel, sheet_name="codigos_territoriales")
df_comunas = df_codigos[df_codigos['División Política Administrativa'] == 'Comuna'].copy()
df_comunas = df_comunas.rename(columns={
    'Código territorial': 'comuna',
    'Territorio': 'nombre_comuna'
})
df_comunas['comuna_norm'] = df_comunas['nombre_comuna'].apply(normalizar_nombre)

# Unir el nombre normalizado al dataset del Censo
df_censo_comunal = df_censo_comunal.merge(
    df_comunas[['comuna', 'nombre_comuna', 'comuna_norm']],
    on='comuna',
    how='left'
)

# Cargar el dataset electoral
df_electoral = pd.read_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\agregado_nacional_comuna_bloque.csv")
df_electoral['comuna_norm'] = df_electoral['Comuna'].apply(normalizar_nombre)

# Verificar que las comunas coinciden
print("=== VERIFICACIÓN DEL CRUCE ===")
print(f"Comunas en Censo (con norm): {df_censo_comunal['comuna_norm'].nunique()}")
print(f"Comunas en electoral (con norm): {df_electoral['comuna_norm'].nunique()}")

# Hacer el merge
df_integrado = df_electoral.merge(
    df_censo_comunal.drop(columns=['comuna']),
    on='comuna_norm',
    how='left'
)

print(f"\n=== DATASET INTEGRADO ===")
print(f"Filas: {len(df_integrado)}")
print(f"Columnas: {df_integrado.columns.tolist()}")

print("\n=== PRIMERAS 5 FILAS ===")
print(df_integrado.head(5).to_string())

print("\n=== NULOS EN VARIABLES DEL CENSO ===")
cols_censo = ['pct_catolicos', 'pct_evangelicos', 'pct_sin_religion', 
              'pct_superior', 'pct_secundaria', 'pct_basica_o_menos', 
              'promedio_escolaridad', 'tasa_desocupacion', 'pct_rural']
print(df_integrado[cols_censo].isnull().sum())
```
