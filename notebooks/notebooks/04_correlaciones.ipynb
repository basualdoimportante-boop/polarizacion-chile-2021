```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr

# Cargar dataset a nivel comuna
df_comuna = pd.read_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\dataset_comuna.csv")

# Variables para correlacionar
variables = [
    'Laakso_Taagepera',
    'pct_catolicos_limpio',
    'pct_evangelicos_limpio',
    'pct_sin_religion_limpio',
    'pct_superior',
    'pct_secundaria',
    'pct_basica_o_menos',
    'promedio_escolaridad',
    'tasa_desocupacion',
    'pct_rural',
]

# Matriz de correlaciones de Spearman
corr_spearman = df_comuna[variables].corr(method='spearman')

print("=== CORRELACIONES CON LAAKSO-TAAGEPERA (Spearman) ===")
print(corr_spearman['Laakso_Taagepera'].sort_values(ascending=False).to_string())

# Significancia estadística
print("\n=== CORRELACIÓN CON SIGNIFICANCIA (Spearman) ===")
print(f"{'Variable':<30} {'r':>8} {'p-valor':>12} {'N':>5}")
print("-" * 60)

for var in variables:
    if var == 'Laakso_Taagepera':
        continue
    subset = df_comuna[['Laakso_Taagepera', var]].dropna()
    r, p = spearmanr(subset['Laakso_Taagepera'], subset[var])
    sig = "***" if p < 0.001 else "**" if p < 0.01 else "*" if p < 0.05 else ""
    print(f"{var:<30} {r:>8.3f} {p:>12.4f} {len(subset):>5} {sig}")
```

```python
import pandas as pd
import glob
import os

archivos = sorted(glob.glob(r"C:\Users\Martín\Desktop\POLARIZACOIN\datos\Diputados\*.xlsx"))
print(f"Archivos: {len(archivos)}")

dataframes = []
for i, archivo in enumerate(archivos, 1):
    xls = pd.ExcelFile(archivo)
    for hoja in xls.sheet_names:
        df_temp = pd.read_excel(archivo, sheet_name=hoja)
        dataframes.append(df_temp)
    print(f"  [{i}/16] {os.path.basename(archivo)}")

df_serval = pd.concat(dataframes, ignore_index=True)
print(f"\nFilas totales: {len(df_serval)}")

# Limpiar filas basura
df_serval = df_serval.dropna(subset=['Comuna', 'Distrito', 'Nro. Mesa'])

# Separar por tipo de voto
mask_nulos = df_serval['Candidato'] == 'VOTOS NULOS'
mask_blancos = df_serval['Candidato'] == 'VOTOS EN BLANCO'
mask_validos = (~mask_nulos) & (~mask_blancos) & (df_serval['Votos TRICEL'].notna())

# Agregar por comuna
nulos = df_serval[mask_nulos].groupby('Comuna')['Votos TRICEL'].sum()
blancos = df_serval[mask_blancos].groupby('Comuna')['Votos TRICEL'].sum()
validos = df_serval[mask_validos].groupby('Comuna')['Votos TRICEL'].sum()

df_participacion = pd.DataFrame({
    'Votos_Nulos': nulos,
    'Votos_Blancos': blancos,
    'Votos_Validos': validos,
}).fillna(0)

df_participacion['Total_Votos'] = df_participacion.sum(axis=1)
df_participacion['pct_nulos'] = df_participacion['Votos_Nulos'] / df_participacion['Total_Votos'] * 100
df_participacion['pct_blancos'] = df_participacion['Votos_Blancos'] / df_participacion['Total_Votos'] * 100
df_participacion['pct_nulos_blancos'] = df_participacion['pct_nulos'] + df_participacion['pct_blancos']

print("\n=== ESTADÍSTICAS ===")
print(df_participacion[['pct_nulos', 'pct_blancos', 'pct_nulos_blancos']].describe())

# Guardar
df_participacion.to_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\participacion_comuna.csv", encoding='utf-8-sig')
print("\nGuardado: resultados/participacion_comuna.csv")
print(f"Comunas: {len(df_participacion)}")
```

```python
import pandas as pd
import unicodedata
import re
from scipy.stats import spearmanr

def normalizar_nombre(nombre):
    if pd.isna(nombre):
        return None
    n = str(nombre).upper().strip()
    n = re.sub(r'\(.*?\)', '', n)
    n = ''.join(c for c in unicodedata.normalize('NFD', n) if unicodedata.category(c) != 'Mn')
    n = n.replace("O'HIGGINS", "OHIGGINS").replace("O HIGGINS", "OHIGGINS")
    n = n.replace("MARCHIGUE", "MARCHIHUE")
    return re.sub(r'\s+', ' ', n).strip()

# Cargar participación (el índice es la Comuna)
df_part = pd.read_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\participacion_comuna.csv", index_col=0)
df_part = df_part.reset_index().rename(columns={'index': 'Comuna'})
df_part['comuna_norm'] = df_part['Comuna'].apply(normalizar_nombre)

# Cargar dataset comuna
df_comuna = pd.read_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\dataset_comuna.csv")

# Cruzar
df_cruce = df_part.merge(
    df_comuna[['comuna_norm', 'promedio_escolaridad', 'pct_superior', 'pct_secundaria',
               'pct_basica_o_menos', 'pct_rural', 'tasa_desocupacion',
               'pct_catolicos_limpio', 'pct_evangelicos_limpio', 'pct_sin_religion_limpio']],
    on='comuna_norm', how='left'
)

print(f"Comunas cruzadas: {len(df_cruce)}")

# Correlaciones
variables = [
    'promedio_escolaridad', 'pct_superior', 'pct_secundaria', 'pct_basica_o_menos',
    'pct_rural', 'tasa_desocupacion',
    'pct_catolicos_limpio', 'pct_evangelicos_limpio', 'pct_sin_religion_limpio'
]

dependientes = ['pct_nulos', 'pct_blancos', 'pct_nulos_blancos']

for dep in dependientes:
    print(f"\n=== {dep} vs VARIABLES ===")
    print(f"{'Variable':<28} {'r':>8} {'p':>10} {'N':>5}")
    print("-" * 55)
    for var in variables:
        subset = df_cruce[[dep, var]].dropna()
        if len(subset) < 10:
            continue
        r, p = spearmanr(subset[dep], subset[var])
        sig = "***" if p<0.001 else "**" if p<0.01 else "*" if p<0.05 else ""
        print(f"{var:<28} {r:>8.3f} {p:>10.4f} {len(subset):>5} {sig}")

# Guardar
df_cruce.to_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\participacion_educacion.csv", index=False, encoding='utf-8-sig')
print("\nGuardado: resultados/participacion_educacion.csv")
```

```python
import pandas as pd

ruta_casen = r"C:\Users\Martín\Desktop\POLARIZACOIN\datos\socioeconomico\SAE_ingresos_2024.xlsx"

# Leer sin encabezado (header=None)
df_raw = pd.read_excel(ruta_casen, sheet_name=0, header=None)

# La fila 1 (índice 1) tiene los nombres reales de las columnas
df_casen = df_raw.iloc[2:].copy()  # Datos desde la fila 2
df_casen.columns = df_raw.iloc[1].tolist()  # Nombres desde la fila 1
df_casen = df_casen.reset_index(drop=True)

# Quedarnos solo con filas que tienen código numérico válido (elimina notas al pie)
df_casen = df_casen[pd.to_numeric(df_casen['Código'], errors='coerce').notna()].copy()
df_casen['Código'] = df_casen['Código'].astype(int)

# Renombrar columnas para uso posterior
df_casen = df_casen.rename(columns={
    'Código': 'cod_com',
    'Región': 'region',
    'Nombre comuna': 'nombre_comuna',
    'Porcentaje de personas en situacion de pobreza de ingresos 2024': 'tasa_pobreza_ingresos_2024',
    'Límite inferior': 'ic_inferior',
    'Límite superior': 'ic_superior',
    'Presencia de la comuna en la muestra Casen': 'en_muestra_casen',
    'Tipo de estimación SAE (***)': 'tipo_estimacion',
})

print("=== FORMA FINAL ===")
print(df_casen.shape)

print("\n=== COLUMNAS ===")
print(df_casen.columns.tolist())

print("\n=== PRIMERAS 5 FILAS ===")
print(df_casen[['cod_com', 'region', 'nombre_comuna', 'tasa_pobreza_ingresos_2024', 'tipo_estimacion']].head().to_string())

print("\n=== COMUNAS ÚNICAS ===")
print(f"Comunas: {df_casen['cod_com'].nunique()}")

print("\n=== TIPO DE ESTIMACIÓN ===")
print(df_casen['tipo_estimacion'].value_counts(dropna=False))

# Guardar
df_casen.to_csv(r"C:\Users\Martín\Desktop\POLARIZACOIN\resultados\casen_ingresos_limpio.csv", index=False, encoding='utf-8-sig')
print("\nGuardado: resultados/casen_ingresos_limpio.csv")
```
